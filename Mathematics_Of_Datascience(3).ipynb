```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set()
```

Plotting  Vector;

```python
plt.quiver(0,0,4,5)
plt.show()
```

```python
plt.quiver(0,0,4,5,scale_units='xy',angles='xy',scale=1,color='b')
plt.xlim(-10,10)
plt.ylim(-10,10)
plt.show()
```

## Adding Vectors;

```python
v_1= np.asarray([0,0,2,5])
v_2=np.asarray([0,0,3,-5])
```

```python
v_3= v_1+v_2
v_3
```

```python
plt.quiver(0,0,2,5,scale_units='xy',angles='xy',scale=1,color='g')
plt.quiver(0,0,3,-5,scale_units='xy',angles='xy',scale=1,color='b')
plt.quiver(0,0,5,0,scale_units='xy',angles='xy',scale=1,color='r')
plt.xlim(-10,10)
plt.ylim(-10,10)
plt.show()
```

## Subtracting Vectors;

```python
v_1 =np.asarray([0,0,2,-5])
v_2 =np.asarray([0,0,3,5])
```

```python
v_3=v_1-v_2
v_3
```

```python
plt.quiver(0,0,2,-5,scale_units='xy',angles='xy',scale=1,color='g')
plt.quiver(0,0,3,5,scale_units='xy',angles='xy',scale=1,color='r')
plt.quiver(0,0,-1,-10,scale_units='xy',angles='xy',scale=1,color='b')
plt.xlim(-10,10)
plt.ylim(-10,10)
plt.show()
```

## Vector Multiplication

```python
v_1 =np.asarray([0,0,2,-5])
v_2 =np.asarray([0,0,3,5])
```

```python
v_3=v_1*v_2
v_3
```

```python
plt.quiver(0,0,2,-5,scale_units='xy',angles='xy',scale=1,color='g')
plt.quiver(0,0,3,5,scale_units='xy',angles='xy',scale=1,color='r')
plt.quiver(0,0,6,-25,scale_units='xy',angles='xy',scale=1,color='b')
plt.xlim(-10,10)
plt.ylim(-10,10)
plt.show()
```
