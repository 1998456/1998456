# DATAFRAME

# Looking at your dataset, using tail() & head() function;


```r
# Creating a Dataframe;
planets <- c("Mercury", "Venus", "Earth", "Mars", "Jupiter", "Saturn", "Uranus", "Neptune")
type <- c("Terrestrial planet", "Terrestrial planet", "Terrestrial planet",
          "Terrestrial planet", "Gas giant", "Gas giant", "Gas giant", "Gas giant")
diameter <- c(0.382, 0.949, 1, 0.532, 11.209,3.4,5.4,0.6)
rotation <- c(58.64, -243.02, 1, 1.03, 0.41,8.9,7.6,6.5)
rings <- c(FALSE, FALSE, FALSE, FALSE, TRUE, TRUE, TRUE, TRUE)
```

```r
planets_df <- data.frame(planets, diameter, rotation, rings)
head(planets_df)
```

```r
# Gaining more insights in the Dataset;
str(planets_df)
```

```r
# Selection Of Dataframe Element;
planets_df[1:3,"type"]
```

```r
planets_df[1:5,"diameter"]
```

```r
# Selecting all Elements Of the Variable Diameter;METHOD(1)
planets_df[,3]
planets_df[,'diameter']
```

```r
# Selecting all Elements Of the Variable Diameter;METHOD(2)
planets_df$diameter
```

## Conditional Selection

```r
# Selecting Planets With Rings;
planets_df[rings==TRUE,]
```

```r
# Performing Conditional Selection using Subset() function;
subset(planets_df,subset=rings)

# OR
subset(planets_df,subset=rings==TRUE)
```

```r
# Selecting planets with diameter less than one;
subset(planets_df,subset=diameter<1)
```

## Sorting a Dataframe;


```r
# Using order() Function;
a <-  c(100,10,1000)
order(a)
```

```r
a[order(a)]
```

```r
# Sorting Planets in a decreasing Order;
positions <- order(planets_df$diameter, decreasing=TRUE)
planets_df[positions,]
```

# LIST

```r
## Creating a List;
# Movie;
mov <- 'Shining'

# Act;
act <- c('jack','kane','kelvin','Ruth','kenny')

# Review;
rev <- data.frame('scores'= c(4.2,3.4,5.6),sources= c('imb1','imb2','imb3'),'reviews'= c('The movie seems great','it was scary though','i will say it was a bit funny'))

shining_list <- list(movie=mov,actor=act,reviews=rev)
shining_list





```

# Selecting Elements In a Dataframe

```r
# Selecting 1st element from 2nd Row;
shining_list[[2]][1]
```

```r
# Selecting 1st,2nd & 3rd elements from 2nd column;
shining_list[[2]][1:3]
```

```r
# Selecting Components;
shining_list[[3]]
shining_list$review
shining_list[[3]][2:3,]
```
