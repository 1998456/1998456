# Comparing Vectors

```r
# Linkedin & Facebook Profile views vectors in a week;
linkedin <- c(16,9,13,5,2,17,14)
facebook <- c(17,7,5,16,8,13,14)
```

```r
# which days did the number of linkedin profiles exceed 15;
linkedin > 15
```

```r
# when was your linkedin profile viewed 5times or less;
linkedin <= 5
```

```r
# When was your linkedin profile visited more often than your facebook profile;
linkedin > facebook
```

# Comparing Matrices

```r
views= matrix(c(linkedin,facebook) , nrow= 2, byrow= TRUE)
views
```

```r
# When are the views less than 14
views <= 14
```
