Random Matrix Subtraction :

```python

# Subtracting Matrix of similar shape:
m4= np.random.randint(10,size=(3,3))
m5= np.random.randint(10,size=(3,3))
```

```python
subt= np.subtract(m4,m5)
```

```python
subt
```

```python
# Subtracting Matrix of Different Shape:
m4= np.random.randint(10,size=(3,3))
m5= np.random.randint(12,size=(3,3))
```

```python
subt= np.subtract(m4,m5)
```

```python
print(subt)
```

```python
# Subtracting Matrix Of Different Shape:
m4= np.random.randint(10,size=(3,3))
m5= np.random.randint(10,size=(4,3))
```

```python
subt2= np.subtract(m4,m5)
```

```python
print(subt2)
```
