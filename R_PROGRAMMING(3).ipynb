## Analyzing Matrices

```r
# Box Office Star Wars;
new_hope <- c(460,314)
empire_strikes <- c(290,247)
return_jedi <- c(309,165)
```

```r
box_office <- c(new_hope, empire_strikes, return_jedi)
box_office
```

```r
star_wars_matrix= matrix(box_office,byrow=TRUE, nrow=3)
```

```r
star_wars_matrix
```

Naming a Matrix

```r
# Countries Region & Title;
region <- c("US", "non-US")
title <- c("New Hope", "Empire Strikes", "Return of the Jedi")
```

```r
colnames(star_wars_matrix) <- region
rownames(star_wars_matrix) <- title
```

```r
star_wars_matrix
```

## Matrix Calculations

```r
# Row Totals;using rowSums()function;
worldwide_vectors <- rowSums(star_wars_matrix)
worldwide_vectors
```

Adding Columns & Rows to a Matrix

```r
# Using cbind() function;
all_wars_matrix <- cbind(star_wars_matrix,worldwide_vectors)
all_wars_matrix
```

```r
total_rev_vector <- colSums(star_wars_matrix)
total_rev_vector
```

```r
# Using cbind() function;
all_wars_matrix1 <- cbind(all_wars_matrix,total_rev_vector)
all_wars_matrix1
```

## Selection Of Matrix Elements

```r
# Select the Non Us revenue for all movies;
non_us_all <- all_wars_matrix1[,2]

# Average Of non_us_revenue;
mean(non_us_all)
```

```r
print(non_us_all)
```

```r
# Select the non_us_revenue for first two movies;
non_us_some <- all_wars_matrix[1:2,2]

# Averge non-us revenue for first two movies;
mean(non_us_some)
```
