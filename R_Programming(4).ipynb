# FACTOR

```r
# Statistial Datatype used to store categorical variables;Use factor()function;
gender_vector <- c('Male', 'Female', 'Female', 'Male', 'Male')
gender_factor <- factor(gender_vector)
gender_factor
```

## Categorical Variables & Factor Levels

```r
# Norminal Variable- has no order;
animal_vec <- c('Dog', 'Rabbit', 'Cat', 'Dog', 'Cat')

# Ordinal_Variable- has natural ordering;
temperature_vec <- c('High','Low','High','Low','Medium')
```

```r
animal_factor <- factor(animal_vec)
animal_factor
```

```r
temperature_factor <- factor(temperature_vec)
temperature_factor
```

## Factor Level

```r
# using the function level();
survey_vec <- c('Male', 'Female', 'Female', 'Male', 'Male')
survey_vec <- factor(survey_vec)
survey_vec
```

```r
level_vec <- levels(survey_vec)
```

```r
level_vec
```

```r
factor_survey_vec_inv <- factor(survey_vec)
factor_survey_vec_inv
```

```r
levels(factor_survey_vec_inv) <- c('Male','Female')
factor_survey_vec_inv
```

```r
survey_vec
```

## Summarizing a Factor

```r
# Using summary() function;
summary(survey_vec)
summary(factor_survey_vec_inv)
```
