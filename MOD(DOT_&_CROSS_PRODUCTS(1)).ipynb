Import Libraries

```python
import numpy as np
```

```python
a= np.array([2,3])

b= np.array([4,5])

a_dot_b = np.dot(a,b)
a_dot_b
```

CROSS Product Of (2) Vectors;

```python
a= np.array([2,3])

b= np.array([4,4])

a_dot_b= np.cross(a,b)
```

```python
print(a_dot_b)
```

```python
c= np.array([6,7,8])

d= np.array([10,4,8])

c_dot_d= np.cross(c,d)
```

```python
print(c_dot_d)
```

Projection Of 'a' Vector on 'V' VECTOR

```python
a= np.array([7,8])

v= np.array([10,11])

# Magnitude Of 'v' Vector;
magnitude_of_v= np.sqrt(sum(v**2))

proj_of_a_on_v=(np.dot(a,v)/magnitude_of_v**2)*v

print("Projection of 'a' vector on 'v' vector = ",proj_of_a_on_v)
```
