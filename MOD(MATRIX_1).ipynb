Creating Matrix With Random Values;

```python
rndmatrix=np.random.rand(3,2)
print(rndmatrix)
```

Creating Matrix With Random Integers;

```python
rndintegers=np.random.randint(100,size=(4,5))
print(rndintegers)
```

Matrix With All Values As 1;

```python
m3=np.ones((2,3))
print(m3)
```

```python
m3=np.ones((2,3),dtype=int)
print(m3)
```

```python
m3=np.ones((10,10),dtype=int)
print(m3)
```

Creating Null / Zero Matrix

```python
nullmatrix=np.zeros((4,4))
print(nullmatrix)
```

```python
nullmatrix=np.zeros((7,7))
print(nullmatrix)
```

Identity Matrix

```python
identity_matrix= np.eye(3,3)
print(identity_matrix)
```

```python
identity_matrix= np.eye(6,6)
print(identity_matrix)
```

Transpose Of Matrix

```python
a= np.random.randint(100,size=(4,5))
print(a)
```

```python
transpose_of_a= np.transpose(a)
print(transpose_of_a)
```
